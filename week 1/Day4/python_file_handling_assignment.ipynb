# file handling assignment

```python
import csv
import json
```

```python
products = [
    [101, "Laptop", "Electronics", 75000, 10],
    [102, "Mouse", "Accessories", 800, 50],
    [103, "Keyboard", "Accessories", 1500, 30]
]
```

## Problem 1 — Write Products into CSV File
Create products.csv and write header and product data using csv.writer().


```python
with open("products.csv", "w", newline="") as file: # creating file with mode without any extra blank line

    writer = csv.writer(file)# it will write python data into the csv fiile

    writer.writerow(["id", "name", "category", "price", "quantity"]) # header rows/ csv file columns name

    writer.writerows(products) # writing all data and data saved

print("CSV File Created Successfully")
```

## Problem 2 — Read Data from CSV File
Read all products from CSV file and display each row.


```python
with open("products.csv", "r") as file: # reading the data which is in file

    reader = csv.reader(file) # read the row of csv

    for row in reader: # reading in loop
        print(row) # current row will be print
```

## Problem 3 — Delete Product from CSV File
Delete product where id == 103 and rewrite updated data into CSV file.


```python
rows = [] # empty list for storing the data which is left after delete

with open("products.csv", "r") as file: # open file in reading mode

    reader = csv.DictReader(file) # all rows will be read in dictornary form

    for row in reader:# loop to read row
        # if 103 data is not matching or there then just store data into list
        if row["id"] != "103":
            rows.append(row)

with open("products.csv", "w", newline="") as file: # opening file in writing mode

    fieldnames = ["id", "name", "category", "price", "quantity"] # define column name

    writer = csv.DictWriter(file, fieldnames=fieldnames) # then the file is rewritten using DictWriter, 
    #excluding the matching record, effectively deleting that product from the CSV file.

    writer.writeheader()

    writer.writerows(rows)

print("Product Deleted")
```

## Problem 4 — Display Products with Price Greater than 5000
Display all products whose price is greater than 5000.


```python
with open("products.csv", "r") as file: # opening the file in read mode

    reader = csv.DictReader(file)# read csv data into the disctionary form

    for row in reader: # all row will read in loop

        if int(row["price"]) > 5000: # while reaading the if the condition is match then only print

            print(row)
```

```python
products = [
    {"id": 101, "name": "Laptop", "category": "Electronics", "price": 75000, "quantity": 10},
    {"id": 102, "name": "Mouse", "category": "Accessories", "price": 800, "quantity": 50},
    {"id": 103, "name": "Keyboard", "category": "Accessories", "price": 1500, "quantity": 30},
    {"id": 104, "name": "Mobile", "category": "Electronics", "price": 25000, "quantity": 15},
    {"id": 105, "name": "Headphones", "category": "Accessories", "price": 2000, "quantity": 20}
]
```

## Problem 5 — Write Products into JSON File
Create products.json and write product data using json.dump().


# json.dump(products, file, indent=4)

products → Data which we have to save
file →in which file we have to save
indent=4 →data will be in readable form by spacing

```python
with open("products.json", "w") as file: # json file open in write mode 

    json.dump(products, file, indent=4)

print("JSON File Created Successfully")
```

## Problem 6 — Read Data from JSON File
Read and display all products from JSON file.


```python
with open("products.json", "r") as file:

    data = json.load(file) # json data will read in python obeject like in dictionary 

    for product in data:
        print(product)
```

## Problem 7 — Delete Product from JSON File
Delete product where id == 102.


```python
with open("products.json", "r") as file: # open the json fle in readble form

    data = json.load(file) #  json data will read in python obeject like in dictionary 

updated_data = [] # remain data after deleting 

for product in data:
    # check if id is 102 or not if yes then add the data into list
    if product["id"] != 102:
        updated_data.append(product)

with open("products.json", "w") as file: # open file in write mode and old data will overwrite

    json.dump(updated_data, file, indent=4) # sving updated data into json

print("Product Deleted")
```

## Problem 8 — Update Product Quantity
Update quantity of Laptop to 25.


```python
with open("products.json", "r") as file: # open in read mode

    data = json.load(file) # json data will read into the python oject like in list and dictionary

for product in data: # check the data in loop for each row
    # if name is match the uppdate the quantity of product
    if product["name"] == "Laptop":

        product["quantity"] = 25

with open("products.json", "w") as file: # rewrite into json file

    json.dump(data, file, indent=4)

print("Quantity Updated")
```

## Problem 10 — Append New Product to CSV
Append a new Monitor product into CSV file.


```python
new_product = [106, "Monitor", "Electronics", 18000, 12] # the data which we want to append

with open("products.csv", "a", newline="") as file: # open the csv file in append mode 

    writer = csv.writer(file)

    writer.writerow(new_product) # write the new_product in row 

print("Product Added")
```

## Problem 11 — Append New Product to JSON
Append the same product into JSON file.


```python
new_product = {
    "id": 106,
    "name": "Monitor",
    "category": "Electronics",
    "price": 18000,
    "quantity": 12
}

with open("products.json", "r") as file:

    data = json.load(file)

data.append(new_product)

with open("products.json", "w") as file:

    json.dump(data, file, indent=4)

print("Product Added")
```

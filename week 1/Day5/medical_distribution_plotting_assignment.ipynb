# Medical Distribution System - Plotting Assignment

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create dataset

medical_data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Medicine_Distributed': [1200, 1500, 1800, 1700, 2100, 2500],
    'Hospital_Orders': [30, 45, 50, 48, 60, 75],
    'Delivery_Time': [5, 4, 6, 5, 3, 2],
    'Revenue': [50000, 65000, 72000, 70000, 85000, 95000]
}

df = pd.DataFrame(medical_data) # help to convert data which in json into the table format

print(df)
```

# Line Chart Problems
1.	Create a line chart showing medicine distribution over months.



```python
plt.figure(figsize=(8,5))

plt.plot(df['Month'],
         df['Medicine_Distributed'],
         marker='o') # markers

plt.title("Medicine Distribution Over Months")
plt.xlabel("Month")
plt.ylabel("Medicine Distributed")
plt.grid(True)

plt.savefig("Medicine_Distribution_Line.png")
plt.show()
```

# 2.	Create a line chart for monthly revenue growth.

```python
plt.figure(figsize=(8,5)) # size chart

plt.plot(df['Month'],
         df['Revenue'],
         marker='o') # marker

plt.title("Monthly Revenue Growth")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)

plt.savefig("Revenue_Growth_Line.png") #saving img
plt.show()
```

# 3.	Create multiple line charts for Medicine_Distributed and Hospital_Orders.
# 4.	Add markers, grid, title, and axis labels to the charts.


```python
plt.figure(figsize=(8,5)) # size of chart

plt.plot(df['Month'],
         df['Medicine_Distributed'],
         marker='o',
         label='Medicine Distributed') # Labeled

plt.plot(df['Month'],
         df['Hospital_Orders'],
         marker='s',
         label='Hospital Orders')

plt.title("Medicine Distribution vs Hospital Orders")
plt.xlabel("Month")
plt.ylabel("Count")
plt.legend()
plt.grid(True)

plt.savefig("Multiple_Line_Chart.png")
plt.show()
```

# Bar Chart Problems
5.	Create a vertical bar chart for hospital orders by month.




```python
plt.figure(figsize=(8,5))

plt.bar(df['Month'],
        df['Hospital_Orders']) # vertical bar method 

plt.title("Hospital Orders by Month")
plt.xlabel("Month") # Xtitle
plt.ylabel("Orders") #ytitle

plt.savefig("Hospital_Orders_Bar.png")
plt.show()
```

# 6.	Create a horizontal bar chart for revenue by month.

```python
plt.figure(figsize=(8,5)) # define figure size

plt.barh(df['Month'],
         df['Revenue'])# horizontal bar chart

plt.title("Revenue by Month")
plt.xlabel("Revenue")
plt.ylabel("Month")

plt.savefig("Revenue_Horizontal_Bar.png")
plt.show()
```

# 7.	Create a grouped bar chart comparing Medicine_Distributed and Hospital_Orders.
# 8.	Add labels, legends, and chart titles.

```python
x = np.arange(len(df['Month'])) #Jan=0, Feb=1, Mar=2, Apr=3, May=4, Jun=5
width = 0.35

plt.figure(figsize=(10,5))

plt.bar(x - width/2,
        df['Medicine_Distributed'],
        width,
        label='Medicine Distributed')


plt.bar(x + width/2,
        df['Hospital_Orders'],
        width,
        label='Hospital Orders')


plt.xticks(x, df['Month']) # Replace numeric values (0,1,2,3,4,5) with actual month names on X-axis

plt.title("Medicine Distribution vs Hospital Orders")
plt.xlabel("Month")
plt.ylabel("Values")
plt.legend() # show the name of bar with color defines

plt.savefig("Grouped_Bar_Chart.png")
plt.show()
```

# Pie Chart Problems
13.	Create a pie chart showing medicine distribution percentage by month.



```python
plt.figure(figsize=(8,8))

plt.pie(df['Medicine_Distributed'],
        labels=df['Month'],
        autopct='%1.1f%%')

plt.title("Medicine Distribution Percentage")

plt.savefig("Medicine_Distribution_Pie.png")
plt.show()
```

# 14.	Create a pie chart for revenue contribution by month.

```python
plt.figure(figsize=(8,8))

plt.pie(df['Revenue'],
        labels=df['Month'],
        autopct='%1.1f%%')

plt.title("Revenue Contribution by Month")

plt.savefig("Revenue_Pie.png")
plt.show()
```

# 15.	Highlight the highest revenue month using explode.
# 16.	Display percentage values in the pie chart.

```python
# Find highest revenue month
max_index = df['Revenue'].idxmax() # from revenue column we are getting the max of index value

explode = [0,0,0,0,0,0] # slice value of pie chart which is normal
explode[max_index] = 0.2 # if the slice is max index the slice length will be 0.2

plt.figure(figsize=(8,8))

plt.pie(df['Revenue'],
        labels=df['Month'], # show month name
        autopct='%1.1f%%', # showing % value
        explode=explode) # max index slice

plt.title("Revenue Contribution (Highest Revenue Highlighted)")

plt.savefig("Revenue_Pie_Explode.png")
plt.show()
```
